```python
# LABEL ENCODING

label_df = df.copy()

label_encoder = LabelEncoder()

for col in label_df.columns:

    if label_df[col].dtype == 'object':

        label_df[col] = label_encoder.fit_transform(label_df[col])

print("Label Encoding Completed.")

display(label_df.head())
```

```python
# ONE HOT ENCODING

onehot_df = pd.get_dummies(df, drop_first=True)

print("One-Hot Encoding Completed.")

print("Shape :", onehot_df.shape)

display(onehot_df.head())
```

```python
# FEATURE SELECTION

# Features (Independent Variables)
X = label_df.drop('Class', axis=1)

# Target Variable (Dependent Variable)
y = label_df['Class']

print("Feature Matrix (X) Shape:", X.shape)
print("Target Variable (y) Shape:", y.shape)

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(y.head())
```

```python
# TRAIN TEST SPLIT

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Set:", X_train.shape)
print("Testing Set :", X_test.shape)
```

```python
# FEATURE SCALING

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature Scaling Completed.")
```

```python
# MODEL EVALUATION FUNCTION

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

results = []

def evaluate_model(model_name, model, X_test, y_test):

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test,
        y_pred,
        average='weighted'
    )

    recall = recall_score(
        y_test,
        y_pred,
        average='weighted'
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average='weighted'
    )

    # ROC-AUC (Multi-Class)
    if hasattr(model, "predict_proba"):

        y_prob = model.predict_proba(X_test)

        roc = roc_auc_score(
            y_test,
            y_prob,
            multi_class='ovr',
            average='weighted'
        )

    else:

        roc = np.nan

    cm = confusion_matrix(y_test, y_pred)

    print("="*60)
    print(model_name)
    print("="*60)

    print("Accuracy :", accuracy)
    print("Precision:", precision)
    print("Recall   :", recall)
    print("F1 Score :", f1)
    print("ROC AUC  :", roc)

    print("\nClassification Report\n")

    print(classification_report(y_test, y_pred))

    plt.figure(figsize=(6,5))

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues'
    )

    plt.title(f"{model_name} Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.show()

    results.append([
        model_name,
        accuracy,
        precision,
        recall,
        f1,
        roc
    ])
```

```python
# LOGISTIC REGRESSION

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000)

lr.fit(X_train_scaled, y_train)

evaluate_model(
    "Logistic Regression",
    lr,
    X_test_scaled,
    y_test
)
```

```python
# DECISION TREE

from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    random_state=42
)

dt.fit(X_train, y_train)

evaluate_model(
    "Decision Tree",
    dt,
    X_test,
    y_test
)
```

```python
# RANDOM FOREST

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

evaluate_model(
    "Random Forest",
    rf,
    X_test,
    y_test
)
```

```python
# SUPPORT VECTOR MACHINE

from sklearn.svm import SVC

svm = SVC(
    probability=True,
    random_state=42
)

svm.fit(X_train_scaled, y_train)

evaluate_model(
    "Support Vector Machine",
    svm,
    X_test_scaled,
    y_test
)
```

```python
# GRADIENT BOOSTING

from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(
    random_state=42
)

gb.fit(X_train, y_train)

evaluate_model(
    "Gradient Boosting",
    gb,
    X_test,
    y_test
)
```

```python
# MODEL COMPARISON

comparison = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ]
)

comparison = comparison.sort_values(
    by="Accuracy",
    ascending=False
)

comparison.reset_index(
    drop=True,
    inplace=True
)

print("="*70)
print("MODEL COMPARISON")
print("="*70)

display(comparison)
```

```python
# BEST MODEL

print("="*70)
print("BEST PERFORMING MODEL")
print("="*70)

best_model = comparison.sort_values(
    by="Accuracy",
    ascending=False
).iloc[0]

display(best_model)

print("\nBest Model :", best_model["Model"])
print("Accuracy   :", round(best_model["Accuracy"],4))
```

```python
# ROC CURVES

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc

# Convert target into binary matrix

classes = np.unique(y_test)

y_test_bin = label_binarize(
    y_test,
    classes=classes
)

models = {

    "Logistic Regression":(lr,X_test_scaled),

    "Random Forest":(rf,X_test),

    "Support Vector Machine":(svm,X_test_scaled),

    "Gradient Boosting":(gb,X_test)

}

plt.figure(figsize=(10,8))

for name,(model,data) in models.items():

    probability = model.predict_proba(data)

    fpr,tpr,_ = roc_curve(
        y_test_bin.ravel(),
        probability.ravel()
    )

    roc_auc = auc(fpr,tpr)

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC={roc_auc:.3f})"
    )

plt.plot([0,1],[0,1],'k--')

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curves")

plt.legend()

plt.show()
```

```python
# ALL CONFUSION MATRICES

from sklearn.metrics import confusion_matrix

models = {

    "Logistic Regression":(lr,X_test_scaled),

    "Decision Tree":(dt,X_test),

    "Random Forest":(rf,X_test),

    "Support Vector Machine":(svm,X_test_scaled),

    "Gradient Boosting":(gb,X_test)

}

for name,(model,data) in models.items():

    prediction = model.predict(data)

    cm = confusion_matrix(
        y_test,
        prediction
    )

    plt.figure(figsize=(5,4))

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues'
    )

    plt.title(name)

    plt.xlabel("Predicted")

    plt.ylabel("Actual")

    plt.show()
```
